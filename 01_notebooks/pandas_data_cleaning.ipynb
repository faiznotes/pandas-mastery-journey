```python
import pandas as pd
import numpy as np

# Membuat dataset kotor sebagai simulasi dunia nyata
data_kotor = {
    "Nama": ["Andi", "Budi", "Citra", "Andi", None, "Fajar", "Gita"],
    "Usia": [21, 22, np.nan, 21, 23, 22, np.nan],
    "Kota": ["Kuningan", "  Cirebon ", "Kuningan", "Kuningan", "Majalengka", None, "cirebon"],
    "Gaji": [4500000, 5000000, 3500000, 4500000, np.nan, 4000000, 4800000]
}

df_kotor = pd.DataFrame(data_kotor)
print("--- Data Asli (Kotor) ---")
display(df_kotor)
```

```python
# 3 teknik utama dalam membersihkan data
# 1. mengecek data yang hilang

df_kotor.isnull() #mengecek baris mana saja yang punya data kossong (True berarti kosong)

#menghitung total data kosong di tiap kolom
df_kotor.isnull().sum()
```

```python
# 2. menangani baris kosong (dropna atau fillna)
df_bersih = df_kotor.dropna() #menghapus baris yang ada data kosongnya (dropna)
df_bersih
```

```python
# mengisi data kosong dengan nilai tertentu (fillna)
rata_usia = df_kotor['Usia'].mean()
df_kotor['Usia'] = df_kotor['Usia'].fillna(rata_usia)
df_kotor
```

```python
#menghapus data duplikat (drop_duplicates)

df_tanpa_duplikat = df_kotor.drop_duplicates()
df_tanpa_duplikat
```
