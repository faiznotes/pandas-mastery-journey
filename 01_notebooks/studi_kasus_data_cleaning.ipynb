```python
import pandas as pd
import numpy as np

# 1. Dataset Tantangan
data_tugas = {
    "ID_Siswa": [101, 102, 103, 101, 105, 106, 107],
    "Nama": ["Budi", "Siti", None, "Budi", "Joko", "siti", "Ani"],
    "Nilai": [85, 90, np.nan, 85, 75, np.nan, 95],
    "Kota_Asal": ["Kuningan", " cirebon ", "Majalengka", "Kuningan", "BANDUNG", "cirebon", None]
}

df_tugas = pd.DataFrame(data_tugas)

# --1. Cek Kolom dan Baris yang kosong ---
print("--- Mengecek Kolom yang Kosong ---")
print(df_tugas.isnull().sum())
print("-"*40)

# --2. Mengahapus data duplikat
df_bersih = df_tugas.drop_duplicates().copy()
print("--- Setelah Menghapus Data Duplikat ---")
display(df_bersih)
print("-"*40)

# --3. Mengisi nilai kosong dengan rata-rata
rata_nilai = df_bersih['Nilai'].mean()
df_bersih['Nilai'] = df_bersih['Nilai'].fillna(rata_nilai)

print("--- Setelah Mengisi Nilai Kosong dengan Rata-rata ---")
display(df_bersih)
print("-"*40)

#--4. Pembersihan teks pada Kolom Asal_Kota
df_bersih['Kota_Asal'] = df_bersih['Kota_Asal'].str.strip().str.title()

print("--- hasil Akhir Data yang Sudah Dibersihkan")
display(df_bersih)

```
