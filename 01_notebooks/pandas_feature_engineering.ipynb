```python
"""
Apa itu Feature Engineering?
=> Merupakan seni menciptakan kolom/variabel baru dari data yang sudah bersih
menggunakan domain knowledge (Pemahaman Bisnis/Logika)

Contoh: Kolom Tanggal Transaksi bisa di ekstak menjadi kolom Hari, Bulan, dan tahun.
Kolom Harga dan Jumlah, bisa buat kolom Total_Belanja

jadi Data mentah jarang sekali siap dipakai untuk analisis mendalam. Sebelum dianalisis
kita harus memperas data mentah tersebut agar memunculkan pola tersembunyi yang bernilai tinggi.
"""
```

```python
import pandas as pd

# 1. Dataset Transaksi Penjualan Mentah
data_transaksi = {
    "ID_Transaksi": [1, 2, 3, 4, 5],
    "Pelanggan": ["Andi", "Budi", "Citra", "Dina", "Eko"],
    "Harga_Satuan": [150000, 50000, 1200000, 300000, 75000],
    "Jumlah_Beli": [2, 5, 1, 3, 4],
    "Metode_Bayar": ["QRIS", "Transfer", "CC", "QRIS", "Cash"]
}

df_tx = pd.DataFrame(data_transaksi)
print("--- Data Asli ---")
display(df_tx)

# Feature engineering: 1. Membuat kolom total_Belanja (harga satuan*Jumlah Beli)
df_tx['Total_Belanja'] = df_tx['Harga_Satuan'] * df_tx['Jumlah_Beli']

# Membuat kolom kategorisasi pelanggan (Conditional Binning)
# Misal: Jika Total_Belanja >500k, Sebut 'VIP', Sisanya 'Reguler'ArithmeticError

def tentukan_status_pelanggan(total):
    if total > 500000:
        return 'VIP'
    else:
        return 'Reguler'

#menerapkan fungsi ke dataframe menggunakan apply()
df_tx['Status_Pelanggan'] = df_tx['Total_Belanja'].apply(tentukan_status_pelanggan)

print("\n-- Data Setelah Feature Engineering ---")
display(df_tx)
```

```python
import pandas as pd
import numpy as np

# Data Penjualan Toko Mainan
data_latihan = {
    "ID_Produk": [101, 102, 103, 101, 105, 106],
    "Nama_Barang": ["Lego", "Boneka", None, "Lego", "Mobil-mobilan", "Rubik"],
    "Harga": [150000, 50000, 75000, 150000, 60000, 5000000], # Perhatikan harga Rubik (Outlier ekstrem!)
    "Terjual": [10, 25, 15, 10, 30, 2]
}

df_latihan = pd.DataFrame(data_latihan)
print("=== DATA SEBELUM CLEANING ===")
display(df_latihan)
print("-"*40)

#menghapuus data duplikat dan mengganti nama_barang yg kosong dengan teks 'Tidak DIketahui'ArithmeticError
df_bersih = df_latihan.drop_duplicates().copy()
print("=== SETELAH MENGAHPUS DATA DUPLIKAT ===")
display(df_bersih)

#membuat kolom baru total_omzet
df_bersih['Total_Omzet'] = df_bersih['Harga'] * df_bersih['Terjual']
print("=== MEMBUAT KOLOM BARU TOTAL_OMZET ===")
display(df_bersih)
```
